```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("population.csv")

# Clean column names
df.columns = df.columns.str.strip()

# Convert to numeric
df['2022'] = pd.to_numeric(df['2022'], errors='coerce')

# Remove missing values
df = df.dropna()

# 🔥 KEEP ONLY REAL COUNTRIES (STRICT FILTER)
# Real countries have proper ISO country codes (not aggregates)
df = df[df['Country Code'].str.fullmatch(r'[A-Z]{3}')]

# 🚨 REMOVE ALL AGGREGATES (FINAL GUARANTEE)
df = df[~df['Country Name'].str.contains(
    "income|IDA|IBRD|demographic|Fragile|HIPC|OECD|World|Euro|Asia|Africa|Europe|America",
    case=False, na=False
)]

# Get Top 15 countries
top15 = df[['Country Name', '2022']] \
            .sort_values(by='2022', ascending=False) \
            .head(15)

# 🔍 Check result (IMPORTANT)
print(top15)

# Plot
plt.figure(figsize=(12,8))
sns.barplot(x='2022', y='Country Name', data=top15)

plt.title("Top 15 Most Populous Countries (2022)")
plt.xlabel("Population")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python

df = df[df['Country Code'].str.len() == 3]
df['2022'] = pd.to_numeric(df['2022'], errors='coerce')
df = df.dropna()
plt.figure(figsize=(8,5))
sns.histplot(df['2022'], bins=20, color='blue')

plt.title("Population Distribution (2022)")
plt.xlabel("Population")
plt.ylabel("Number of Countries")

plt.grid(True)
plt.show()
```
