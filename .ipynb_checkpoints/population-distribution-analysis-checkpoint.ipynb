```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="whitegrid")
```

```python
df = pd.read_csv("population.csv")
```

```python
print(df.head())
print(df.shape)
print(df.columns)
```

```python
df.columns = df.columns.str.strip()
```

```python
df['2022'] = pd.to_numeric(df['2022'], errors='coerce')
```

```python
df.dropna(inplace=True)
```

```python
df = df[~df['Country Name'].str.contains(
    "income|IDA|IBRD|World|Europe|Asia|Africa|demographic|Least developed",
    case=False, na=False
)]
```

```python
df_sorted = df.sort_values(by='2022', ascending=False)

top15 = df_sorted[['Country Name', '2022']].head(15)

print(top15)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['2022'],
    bins=20,
    color='skyblue',
    edgecolor='black'
)
plt.title("Population Distribution (2022)", fontsize=14)
plt.xlabel("Population")
plt.ylabel("Number of Countries")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.barplot(
    x='2022',
    y='Country Name',
    data=top15,
    hue='Country Name',   # fix warning
    palette='viridis',
    legend=False
)

plt.title("Top 15 Most Populous Countries (2022)", fontsize=14)
plt.xlabel("Population")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```
